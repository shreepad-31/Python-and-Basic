```python
import numpy as np
```

### Array from list:

```python
arr_1d = np.array([1, 2, 3])
print(arr_1d)

arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print(arr_2d)



pylist = [1, 2, 3]

print("Multiplication of List", pylist * 2)

print("Multiplication of Array", arr_1d * 2)
print("Multiplication of Array", arr_2d * 2)
```

### Matrix from Array:

```python
zeros = np.zeros((3, 4)) 
print("Zero Matrix", zeros)

ones = np.ones((2, 3))
print("Matrix with all entries Ones", ones)

full = np.full((2, 2), 7)
print("Matrix with all entries of given number", full)

random = np.random.random((2, 3))
print("Matrix with Random entries", random)

sequence = np.arange(0, 10, 2) #(start, end + 1, steps)
print("Sequence Array", sequence)
```

### Vector, Matrix and Tensor:

```python
vector = np.array([1, 2, 3])
matrix = np.array([[1, 2, 3], [2, 3, 4]])
tensor = np.array([[[1, 2], [2, 3]],
                   [[3, 4], [4, 5]]])

print("Vector", vector)
print("Matrix", matrix)
print("Tensor", tensor)
```

### Array Properties:

```python
arry = np.array([[1, 2, 3],
                [2, 3, 4]])

print("Order", arry.shape) # Gives Order (n, m)
print("Dimension", arry.ndim) # Gives Dimension
print("Elements", arry.size) # Gives number of elements
```

### Array Reshaping:

```python
arr = np.arange(12) 
reshaped = arr.reshape(3, 4)
print("Reshaped", reshaped)

flatten = reshaped.flatten()
print("Flattened", flatten)

raveled = reshaped.ravel() # Ravel (It returns view instead of copy)
print("Raveled", raveled)

transpose = reshaped.T
print("Transpose", transpose)
```
