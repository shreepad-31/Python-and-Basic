```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

### Kernal Density Estimation Plot

```python
cars = sns.load_dataset("mpg").dropna()
```

```python
cars.shape
```

```python

cars.head()
```

#### Univariate

```python
sns.kdeplot(cars.horsepower, shade = True)
```

```python
sns.kdeplot(cars.horsepower, cumulative = True)
```

#### Bivariate

```python
sns.kdeplot(x= cars.horsepower, y= cars.acceleration)
```

```python
sns.kdeplot(x= cars.horsepower, y= cars.acceleration,shade= True, shade_lowest= False, cbar= True)
```
