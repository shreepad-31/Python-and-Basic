```python
import matplotlib.pyplot as plt
import numpy as np
```

### Graph Plot

```python
x = np.array([2023, 2024, 2025, 2026])
y = np.array([15, 25, 30, 20])

plt.plot(x, y)
plt.show()
```

### Customization of Plot

```python
x = np.array([2023, 2024, 2025, 2026])
y = np.array([15, 25, 30, 20])

plt.title("Class size", fontsize = 25, font = "Arial", fontweight = "bold", color = "g")

plt.xlabel("Year", fontsize = 20, font = "Arial", fontweight = "bold", color = "g") 
plt.ylabel("Students", fontsize = 20, font = "Arial", fontweight = "bold", color = "g")

plt.plot(x, y)
plt.show()
```

```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([17, 23, 38, 5])
y3 = np.array([13, 15, 19, 30])

plt.title("Class size")

plt.xlabel("Year")
plt.ylabel("Students")

plt.tick_params(axis = "both", color = "r")

plt.plot(x, y1) 
plt.plot(x, y2)
plt.plot(x, y3)

plt.xticks(x) # Only ticks at given 'x'

# plt.plot(x, y1, **line_style) # if u have to make some graph customization just make make a dictionary (here it is "line_style") of key-value
# {marker = ".", markersize = 10, markerfacecolor = "#091d19", linestyle = "solid", linewidth = 2, color = "#5186d6"} --> Customizations

plt.show()
```

### Grid Plot

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 11, 9, 25, 21])

plt.xticks(x)

plt.grid(axis = "y", linewidth = 2, color = "lightgrey", linestyle = "dashdot")

plt.plot(x, y)

plt.show()
```

### Bar Graph

```python
categories = np.array(["Netflix", "Prime", "Hotstar", "HBOMAX", "VOOT"])
values = np.array([10, 6, 16, 7, 2])

# plt.barh(categories, values, color = "r") --> Horizontal Bar Graph, remember x-y changes

plt.bar(categories, values, color = "r")

plt.title("Content Consumption")
plt.xlabel("OTTs")
plt.ylabel("Crazy")

plt.show()
```

### Bar Chart

```python
categories = ["Freshers", "SYs", "TYs", "FYs"]
values = np.array([1600, 1200, 1000, 900])
colors = ["red", "green", "blue", "yellow"]

plt.pie(values, labels = categories,
                autopct = "%1.1f%%",
                colors = colors,
                explode = [0, 0, 0, 0.05],
                shadow =  True,
                startangle = 20)

plt.title("COEP Tech")

plt.show()
```

### Scatter Graph

```python
x1 = np.array([0, 1, 1, 2, 3, 4, 5, 7, 7, 8])
y1 = np.array([55, 60, 65, 68, 70, 75, 78, 82, 85, 87])

x2 = np.array([0, 1, 2, 2, 3, 4, 5, 6, 7, 8])
y2 = np.array([50, 58, 65, 70, 72, 78, 83, 92, 95, 97])

plt.scatter(x1, y1, color = "r",
                    alpha = 0.5,
                    s = 100,
                    label = "Class A")

plt.scatter(x2, y2, color = "b",
                    alpha = 0.5,
                    s = 100,
                    label = "Class B")

# Alpha is for tansparency

plt.title("Test Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Grades")

plt.legend()

plt.show()
```

### Histograms 

```python
scores = np.random.normal(loc = 80, scale = 10, size = 100)
scores = np.clip(scores, 0, 100)

plt.hist(scores, bins = 10,
                  color = "lightgreen",
                  edgecolor = "black")
plt.xlabel("Grades")
plt.ylabel("Number of students")
plt.show()
```

### Subplots

Figure : The entire canvas, Ax : A singel plot (subplot)

We can show bar, plot and many other.

```python
figure, axes = plt.subplots(2, 2)

x = np.array([1, 2, 3, 4, 5])

axes[0, 0].plot(x, x*2, color = "lightgreen")
axes[0, 0].set_title("x*2")

axes[0, 1].plot(x, x**2, color = "skyblue")
axes[0, 1].set_title("x**2")

axes[1, 0].plot(x, x**3, color = "purple")
axes[1, 0].set_title("x**3")

axes[1, 1].plot(x, x%2, color = "yellow")
axes[1, 1].set_title("x%2")

plt.show()
```
