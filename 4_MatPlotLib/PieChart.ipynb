- Use case - To find contibution on a standard scale

```python
data = [23,45,100,20,49]
subjects = ['English','Science','Maths','Social Studies','Hindi']
plt.pie(data,labels=subjects)

plt.show()
```

```python
rcb = pd.read_csv('zRCB.csv')
rcb
```

```python
plt.pie(rcb['batsman_runs'] ,labels= rcb['batsman'], autopct= '%0.1f%%', explode=[0, 0, 0, 0, 0, 0.1])

plt.show()
```

```python
color= ['blue','green','yellow','pink','cyan','brown']
plt.pie(rcb['batsman_runs'] ,labels= rcb['batsman'], colors= color)

plt.savefig('zsample.png')
plt.show()
```
