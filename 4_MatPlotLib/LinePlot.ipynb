```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
roko = pd.read_csv("zRoKo.csv")
```

```python
roko.head()
```

Bivariate Analysis:
Categorical -> Numerical and Numerical -> Numerical
// Use case - Time series data

```python
plt.plot(roko['index'], roko['V Kohli'], color= 'red', linewidth= 3, marker= 'o', markersize= 10, label = 'Virat')
plt.plot(roko['index'], roko['RG Sharma'], color= 'blue', linestyle= 'dashdot', label= 'Rohit')

plt.title("Rohit Sharma VS Virat Kohli")

plt.xlabel("Year")
plt.ylabel("Runs")

plt.xlim(2008, 2016)
plt.ylim()

plt.legend(loc = 'lower right')

plt.grid()

plt.show()
```
