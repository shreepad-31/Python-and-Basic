```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

- Bivariate Analysis
- Numerical vs Categorical
- Use case - Aggregate analysis of groups

```python
children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']
```

```python
plt.bar(colors, children, color='black')
plt.show()
```

```python
plt.barh(colors, children, color='black')
plt.show()
```

```python
batsman_season = pd.read_csv('zBatsmanSeason.csv')
batsman_season
```

```python
plt.bar(np.arange(batsman_season.shape[0]) - 0.2, batsman_season['2015'], width= 0.2, color= 'red')
plt.bar(np.arange(batsman_season.shape[0]), batsman_season['2016'], width= 0.2, color= 'orange')
plt.bar(np.arange(batsman_season.shape[0]) + 0.2, batsman_season['2017'], width= 0.2, color= 'yellow')

plt.xticks(np.arange(batsman_season.shape[0]), batsman_season['batsman'], rotation= 'vertical')

plt.show()
```

```python
plt.bar(batsman_season['batsman'], batsman_season['2015'], color= 'red', label= '2015')
plt.bar(batsman_season['batsman'], batsman_season['2016'], color= 'orange', bottom= batsman_season['2015'], label= '2016')
plt.bar(batsman_season['batsman'], batsman_season['2017'], color= 'yellow', bottom= batsman_season['2015'] + batsman_season['2016'], label= '2017')

plt.legend()
plt.show()
```
