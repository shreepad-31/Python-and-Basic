```python
import numpy as np
```

### Advance Operation with Business examples

```python
# Data Structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000], # Paradise Biryani
    [2, 120000, 140000, 160000, 190000], # Beijing Bites
    [3, 200000, 230000, 260000, 300000], # Pizza Hub
    [4, 180000, 210000, 240000, 270000], # Burger Point
    [5, 160000, 185000, 205000, 230000]  # Chai Point
])
```

### Operations over given Data

```python
print("Sales data shape", sales_data.shape)
print(sales_data[:, 1:])

print(np.sum(sales_data, axis = 1))
print(np.sum(sales_data[: ,1:], axis = 1))

print(np.sum(sales_data, axis = 0))
print(np.sum(sales_data[: ,1:], axis = 0))

monthly_avg = sales_data[:, 1:] / 12 # Broadcasting
print(monthly_avg)
```

```python
min_sales = np.min(sales_data[:, 1:], axis = 1) # per Restaurent
print(min_sales)

max_sales = np.max(sales_data[:, 1:], axis = 0) # per Year
print(max_sales)

avg_sales = np.mean(sales_data[:, 1:], axis = 1) # per Restaurent
print(avg_sales)

cummulative_sales = np.cumsum(sales_data[:, 1:], axis = 1)
print(cummulative_sales)
```

### Vector Operations

```python
vector1 = np.array([1, 2, 3, 4, 5])
vector2 = np.array([6, 7, 8, 9, 10])


print("Vector Addition", vector1 + vector2)
print("Vector Multiplication", vector1 * vector2)
print("Dot Product", np.dot(vector1, vector2))

angle = np.arccos(np.dot(vector1, vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print("Angle between Vectors", angle)
```
