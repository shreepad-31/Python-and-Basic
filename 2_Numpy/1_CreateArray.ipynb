```python
import numpy as np
```

```python
# Array
arr_1d = np.array([1, 2, 3])
print(arr_1d)
```

```python
# Matrics
arr_2d = np.array([[1, 2, 3], [2, 3, 4], [3, 4, 5]])
print(arr_2d)
```

```python
# Tensor
arr_3d = np.array([[[1, 2, 3], [2, 3, 4]], [[3, 4, 5], [4, 5, 6]]])
print(arr_3d)
```

```python
# Array of Given Data Type
arr = np.array([1, 2, 3], dtype= float)
print(arr)
```

```python
# Initializing Sequencial Array
arr = np.arange(1, 11, 2);
print(arr)
```

```python
# Reshaping
arr = np.arange(16).reshape(1, 2, 8)
print(arr)
```

```python
# Identity Matrics
arr = np.identity(3, dtype= int)
print(arr)
```

```python
# Array of Ones' and Zeros'
ones = np.ones((2, 2), dtype=int)
print(ones)
zeros = np.zeros((3, 3))
print(zeros)
```

```python
# Initializing Random Array
arr = np.random.random((3, 4))
print(arr)
```

```python
a = np.random.randint(10, 100, (6, 4))
print(a)
```

```python
# LineSpacing Array
arr = np.linspace(-10, 10, 10, dtype= int)
print(arr)
```
