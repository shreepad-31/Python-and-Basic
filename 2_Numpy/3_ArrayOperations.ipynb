```python
import numpy as np
```

### Array Operations

```python
a1 = np.arange(12).reshape((3, 4))
a2 = np.arange(12, 24).reshape((3, 4))
```

```python
# Scaler Operations

# Arithmatic
print(a1 + 2)

print(a1 ** 3)
```

```python
# Relational
print(a1 == 1)
```

```python
# Vector Operations
print(a1 + a2)

print(a1 * a2)
```

```python
a = np.random.randint(10, 100, (3, 4))
print(a)
```

```python
# Boolean Masking
a[(a > 50) & (a % 2 == 0)]
```

```python
a[a % 7 == 0]
```

### Numpy Mathematical Operations

```python
a1 = np.random.random((3, 3))
a1 = np.round(a1 * 100)
a1
```

```python
# Mathematical Operations
print("Sum a1: ", np.sum(a1))

print("Product a1: ", np.prod(a1))

print("Sin of a1: ", np.sin(a1))
```

```python
# Min/Max
print("Max of Row", np.max(a1, axis= 1))

print("Min of Col", np.min(a1, axis= 0))
```

```python
# Mean/Mode/Median/var/std
print("Mean of a1=", np.mean(a1))

print("Median of a1=", np.median(a1))

print("Std of a1=", np.std(a1, axis= 1))

print("Var of a1=", np.var(a1, axis= 1))
```

```python
# Log and Exponent
np.log(a1) 
```

```python
np.exp(a1)
```

### Working with mathematical formulas

```python
# sigmoid
def sigmoid(array):
  return 1/(1 + np.exp(-(array)))


a = np.arange(100)

sigmoid(a)
```

```python
# mean squared error

actual = np.random.randint(1,50,25)
predicted = np.random.randint(1,50,25)
```

```python
def mse(actual,predicted):
  return np.mean((actual - predicted)**2)

mse(actual,predicted)
```
