```python
import numpy as np
```

```python
a1 = np.arange(10)
for i in a1:
    print(i)
```

```python
a2 = np.arange(16).reshape(4, 4)
a2
```

```python
for i in a2:
    print(i)
```

```python
for i in a2.ravel():
    print(i)
```

```python
a3 = np.arange(16).reshape((2, 2, 2, 2))
print(a3)
```

```python
for i in a3:
    print(i)
```

```python
for i in np.nditer(a3):
    print(i)
```
