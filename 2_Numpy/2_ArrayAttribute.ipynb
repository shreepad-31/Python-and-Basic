```python
import numpy as np
```

```python
a1 = np.arange(10, dtype= np.int32)
a2 = np.arange(12, dtype= float).reshape((3, 4))
a3 = np.arange(16).reshape((2, 2, 2, 2))
```

```python
# Some Basic Attributes
print("Dimension of a2:", a2.ndim)

print("Size of a2:", a2.size)

print("Shape of a3:", a3.shape)

print("Itemsize of a1:", a1.itemsize)

print("Itemsize of a3:", a3.itemsize)

print("Datatype of a1:", a1.dtype)
```

```python
# Changing DataTypes of Existing Array
a3_copy = a3.astype(np.int32)
a3_copy.dtype
```
