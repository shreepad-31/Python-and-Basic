## Series: 1D-Data Chart

```python
data = [100, 102, 104, 400, 500]

series = pd.Series(data, index = ["a", "b", "c", "d", "e"])

series.loc["b"] = 200
series.iloc[2] = 300

print(series)

print(series[series > 200])
```

```python
calories = {"Day1" : 1750, "Day2" : 2100, "Day3": 1700}

series = pd.Series(calories)
series.loc["Day3"] += 500

print(series)

print(series[series > 2000])
print(series[series < 2000])
```

## Dataframes : 2-D Data Chart

```python
data = {
    "Name" : ["Danny", "Jon Snow", "Jamie"],
    "Age" : [37, 27, 33],
    }

df = pd.DataFrame(data, index = ["Character 1 ", "Character 2", "Character 3"])

# Adding a new Column
df["Interest"] = ["Throne", "Winterfell", "Cersi"]

# Adding a new Row
new_row = pd.DataFrame([{"Name" : "Ned", "Age" : 53, "Interest" : "Peace"},
                         {"Name" : "Tywin", "Age" : 48, "Interest" : "Power"}],
                         index = ["Character 4", "Character 5"])

df = pd.concat([df, new_row])

print(df)
print("\n", df.loc["Character 2"])
```
