```python
import numpy as np
import pandas as pd
```

### Series Initialization

```python
# Series is 1D Data Structure of Any DataType
country = ['India','Pakistan','USA','Nepal','Srilanka']
pd.Series(country)
```

```python
marks = [67,57,89,100]
subjects = ['maths','english','science','hindi']

report = pd.Series(marks,index=subjects, name= "Marks")
print(report)
```

```python
# Series by Dictionary
marks = {
    'maths':67,
    'english':57,
    'science':89,
    'hindi':100
}

marks_series = pd.Series(marks,name='Marks')
marks_series
```
